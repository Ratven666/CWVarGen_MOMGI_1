```python
import pandas as pd
import numpy as np
import os
```

```python
df_1 = pd.read_csv(os.path.join("Вариант", "Серия_1", "FMWY.txt"), index_col=0)
df_1 = df_1.rename(columns={'X': 'X_0', 'Y': 'Y_0', "Z": "Z_0"})
df_1.index = pd.to_datetime(df_1.index)
df_1
```

```python
df_2 = pd.read_csv(os.path.join("Вариант", "Серия_1", "WLTU.txt"), index_col=0)
df_2 = df_2.rename(columns={'X': 'X_1', 'Y': 'Y_1', "Z": "Z_1"})
df_2.index = pd.to_datetime(df_2.index)
df_2
```

```python
df_3 = pd.concat([df_1, df_2], axis=1)
df_3
```

```python
df_4 = df_3.dropna()
df_4
```

```python
df_4.loc[:, 'dX'] = df_4["X_1"] - df_4["X_0"]
df_4.loc[:, 'dY'] = df_4["Y_1"] - df_4["Y_0"] 
df_4.loc[:, 'dZ'] = df_4["Z_1"] - df_4["Z_0"]
df_4
```

```python
df_5 = df_4[["dX", "dY", "dZ"]]
df_5
```

```python
df_6 = df_4[["dX", "dY", "dZ"]]
df_6['s_dist'] = (df_6["dX"] ** 2 + df_6["dY"] ** 2 + df_6["dZ"] ** 2) ** 0.5
df_6['azimuth'] = np.degrees(np.arctan2(df_6['dY'], df_6['dX']))
df_6['zenith'] = np.degrees(np.arccos(df_6['dZ'] / df_6['s_dist']))
df_6
```

```python
dsc = df_6.describe()[["s_dist", "azimuth", "zenith"]].loc[["mean", "std"]]
dsc
```

```python
dsc.loc[["std"]] = dsc.loc[["std"]].apply(lambda row: row / (len(df_6) ** 0.5), axis=1)
dsc
```

```python
dsc.loc["std", 'azimuth']
```

```python
dsc.loc["std", "azimuth"] = dsc.loc["std", 'azimuth'] * 3600
dsc.loc["std", "zenith"] = dsc.loc["std", "zenith"] * 3600
dsc
```
