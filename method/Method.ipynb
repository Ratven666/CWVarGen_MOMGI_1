```python
import pandas as pd
import numpy as np
import os
```

```python
df_1 = pd.read_csv(os.path.join("Вариант", "Серия_2", "FMWY.txt"), index_col=0)
df_1 = df_1.rename(columns={'X': 'X_0', 'Y': 'Y_0', "Z": "Z_0"})
df_1.index = pd.to_datetime(df_1.index)
df_1
```

```python
df_2 = pd.read_csv(os.path.join("Вариант", "Серия_2", "CLGK.txt"), index_col=0)
df_2 = df_2.rename(columns={'X': 'X_1', 'Y': 'Y_1', "Z": "Z_1"})
df_2.index = pd.to_datetime(df_2.index)
df_2
```

```python
df_3 = pd.concat([df_1, df_2], axis=1).dropna()
df_3
```

```python
df_3.loc[:, 'dX'] = df_3["X_1"] - df_3["X_0"]
df_3.loc[:, 'dY'] = df_3["Y_1"] - df_3["Y_0"] 
df_3.loc[:, 'dZ'] = df_3["Z_1"] - df_3["Z_0"]
```

```python
df_4 = df_3[["dX", "dY", "dZ"]]
df_4
```

```python
df_4 = df_3[["dX", "dY", "dZ"]]
df_4.loc[:, 's_dist'] = (df_4["dX"] ** 2 + df_4["dY"] ** 2 + df_4["dZ"] ** 2) ** 0.5
df_4.loc[:, 'azimuth'] = np.degrees(np.arctan2(df_4['dY'], df_4['dX']))
df_4.loc[:, 'zenith'] = np.degrees(np.arccos(df_4['dZ'] / df_4['s_dist']))
```

```python
dsc = df_4.describe().loc[["mean", "std"]][["s_dist", "azimuth", "zenith"]]
dsc
```

```python
dsc.loc[["std"]] = dsc.loc[["std"]].apply(lambda row: row / (len(df_4) ** 0.5), axis=1)
dsc
```

```python
dsc.loc["std", "azimuth"] = dsc.loc["std", 'azimuth'] * 3600
dsc.loc["std", "zenith"] = dsc.loc["std", "zenith"] * 3600
dsc
```

```python
azimuth = dsc.loc["mean", 'azimuth']
dsc.loc["mean", "azimuth"] = azimuth + 360 if azimuth < 0 else azimuth
dsc
```
